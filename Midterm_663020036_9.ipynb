```python
missing_records_count = merged_df[merged_df['value'].isna()].groupby('station_code').size().reset_index(name='missing_records')
total_records_count = merged_df.groupby('station_code').size().reset_index(name='total_records')

# Combine the counts
missing_summary = pd.merge(
    total_records_count,
    missing_records_count,
    on='station_code',
    how='left'
).fillna(0) # Fill NaN in 'missing_records' with 0 for stations with no missing data

# Calculate proportion of missing records
missing_summary['missing_proportion'] = (missing_summary['missing_records'] / missing_summary['total_records']) * 100

print("Summary of missing records per station:")
display(missing_summary.head())
print("\nStations with the highest proportion of missing records:")
display(missing_summary.sort_values(by='missing_proportion', ascending=False).head())
```

## 5.เติมข้อมูล

> เพิ่ม Day of Year (DOY)

```python
merged_df['dayofyear'] = merged_df['measure_datetime'].dt.dayofyear
```

> คำนวณ Median จาก historical data (แยกตามสถานี + DOY)

```python
median_doy_station = (
    merged_df
    .dropna(subset=['value'])
    .groupby(['station_code', 'dayofyear'])['value']
    .median()
    .reset_index()
    .rename(columns={'value': 'median_value'})
)
```

> นำค่า Median มา merge กลับ

```python
merged_df = pd.merge(
    merged_df,
    median_doy_station,
    on=['station_code', 'dayofyear'],
    how='left'
)
```

> เติมข้อมูลที่สูญหายด้วย Median

```python
merged_df['value_filled'] = merged_df['value'].fillna(merged_df['median_value'])
```

> ตรวจสอบว่ายังมี record หายอีกหรือไม่

```python
remaining_missing = merged_df['value_filled'].isna().sum()
print(f"Remaining missing records after imputation: {remaining_missing}")
```

#6.เหตุใดจึงเลือกใช้ Median แทน Mean

ในการเติมข้อมูลที่สูญหายในงานนี้ เลือกใช้ค่า Median แทนค่า Mean เนื่องจากลักษณะของข้อมูลปริมาณน้ำฝนมีการกระจายที่ไม่สม่ำเสมอ และมักพบค่าที่สูงผิดปกติในบางช่วงเวลา เช่น วันที่เกิดฝนตกหนัก ซึ่งค่าลักษณะนี้สามารถส่งผลกระทบต่อค่าเฉลี่ยได้โดยตรง หากใช้ค่า Mean ค่าที่นำมาเติมอาจสูงหรือต่ำเกินความเป็นจริง และไม่สะท้อนค่าปกติของข้อมูลในช่วงเวลานั้น

ค่า Median เป็นค่ากลางของข้อมูลและมีความทนทานต่อค่าผิดปกติมากกว่า จึงเหมาะสำหรับใช้แทนค่าที่หายไป เพราะช่วยลดอิทธิพลของค่าที่สูงหรือต่ำเกินไป นอกจากนี้ การคำนวณค่า Median จากข้อมูลย้อนหลังโดยอ้างอิงตามวันของปี (Day of Year) และแยกตามสถานี ยังช่วยรักษาลักษณะการเปลี่ยนแปลงตามฤดูกาลของข้อมูล และทำให้ค่าที่เติมมีความสอดคล้องกับพฤติกรรมของข้อมูลในพื้นที่เดียวกัน

การเลือกใช้ Median แทน Mean จึงช่วยให้ข้อมูลหลังการเติมมีความสมเหตุสมผลมากขึ้น ลดการบิดเบือนของข้อมูล และลดผลกระทบต่อการวิเคราะห์ในขั้นตอนถัดไป เช่น การวิเคราะห์แนวโน้มหรือการสร้างแบบจำลอง ทำให้ผลการวิเคราะห์มีความน่าเชื่อถือและสะท้อนสภาพข้อมูลจริงได้ดียิ่งขึ้น
